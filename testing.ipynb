```python
import numpy as np

from cqpsolver import Problem, Solver, SolverState
```

```python
test_Q: np.ndarray = np.array([[1.0, 0.0], [0.0, 1.0]])
test_q: np.ndarray = np.array([[-2.0], [-1.0]])

# -x1 <= 0 and -x2 <= 0
test_G: np.ndarray = np.array([[-1.0, 0.0], [0.0, -1.0]])
test_h: np.ndarray = np.array([[0.0], [0.0]])

# x1 + x2 = 1
test_A: np.ndarray = np.array([[1.0, 1.0]])
test_b: np.ndarray = np.array([[1.0]])

test_problem: Problem = Problem(test_Q, test_q, test_G, test_h, test_A, test_b)

solver: Solver = Solver(test_problem, tol=1e-5)
%timeit state_history: list[SolverState] = solver.solve()
```
